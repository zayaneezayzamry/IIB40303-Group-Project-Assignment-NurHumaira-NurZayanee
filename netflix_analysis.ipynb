# 5. Exploratory Data Analysis

## 5.1 Distribution of Content Type

```python
sns.countplot(data=df, x='type')

plt.title('Movies vs Tv Shows')
plt.xlabel('Content Type')
plt.ylabel('Number of Titles')
plt.show()
```

### Interpretation
Clear Dominance: The storyline highlights a major difference in the platform's collection, demonstrating that films are far more common than TV series.

Large Margin: The bar for movies is more than two times as tall as the bar for TV series, highlighting the platform's strong preference for movie over TV show.

Library Focus: This distribution shows that building a large film archive has traditionally been prioritized over acquiring current television content.

## 5.2 Top Content Producing Countries

```python
df['country'].value_counts().head(10).plot(kind='bar')

plt.title('Top 10 Producing Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')
plt.show()
```

### Interpretation
Overwhelming Leadership: The United States is the clear leader in content production on the platform. Its volume is much higher than all other countries in the dataset.

Secondary Tier: India ranks second and has a clear lead over the remaining countries.

Data Limitation: A large portion of the content is labeled as “Unknown.” This category ranks close to India, which suggests that a significant amount of geographic information is missing or not recorded.

Gradual Decline: After the United Kingdom, the other top content-producing countries, mainly from Asia such as Japan and South Korea, North America such as Canada and Mexico, and Europe such as Spain and France, show a steady decrease in content volume, forming a long tail of smaller contributors.

## 5.3 Content Ratings Distribution

```python
df['rating'].value_counts().plot(kind='bar')

plt.xlabel('Rating')
plt.ylabel('Number of Titles')
plt.show()
```

### Interpretation
Dominance of Mature Content: The plot shows that TV-MA is the most frequent rating on the platform. This indicates a strong concentration of content intended for mature adult audiences.

Strong Teen Presence: TV-14 is the second most common rating and has a clear lead over most other categories. This suggests that a large portion of the content is targeted at teenagers.

Moderate Mid-Tier Ratings: Ratings such as TV-PG, R, and PG-13 form a middle group with moderate and similar representation. These categories mainly include content for older children, families, and general audiences.

Minority and Anomalous Data: Content aimed specifically at young children, such as TV-Y, TV-Y7, and G, makes up a relatively small portion of the library. At the lower end of the chart, there are also a few inconsistencies where values like “74 min” or “84 min” appear, suggesting formatting issues where runtimes were mistakenly classified as ratings.

### 5.4 Release Year Trend

```python
df['release_year'].value_counts().sort_index().plot()

plt.title('Content Release Trend')
plt.xlabel('Year')
plt.ylabel('Number of Titles')
plt.show()
```

### Interpretation
Long Period of Inactivity: For several decades in the mid to late 20th century, content volume remained low and relatively flat. This suggests that only a small number of older or historical titles were available.

Gradual Growth: Around the turn of the century, there is a slight upward trend. This indicates a slow but steady expansion of the library as more modern content began to be added.

Exponential Surge: In the most recent decade, the chart shows a very sharp and dramatic increase. This reflects a period of rapid growth where most of the platform’s content was released.

Recent Drop-off: At the very end of the graph, there is a sudden decline. This is likely due to a data issue, such as incomplete records for the final year or inconsistencies in data collection during that period.

## 5.5 Genre Analysis

```python
genres = df['listed_in'].str.split(', ').explode()

genres.value_counts().head(10).plot(kind='bar')

plt.xlabel('Genre')
plt.ylabel('Number of Titles')
plt.show()
```

### Interpretation
Leading Categories: International Movies and Dramas emerge as the most dominant genres on the platform by a substantial margin, capturing the largest share of the overall content catalog.

Strong Mid-Tier Content: Comedies and International TV Shows form a solid secondary tier of popular offerings, comfortably outpacing the lower-ranked genres.

Global vs. Domestic Focus: The high placement of both International Movies and International TV Shows underscores the platform's heavy reliance on non-domestic or globally sourced content to flesh out its library.

Evenly Distributed Tail: The remaining genres including Documentaries, Action & Adventure, TV Dramas, Independent Movies, Children & Family Movies, and Romantic Movies show a very gradual, uniform decline, representing a stable and balanced variety of niche or specific content categories at the lower end of the top ten.

## 5.6 Content Added to Netflix by Year

```python
df['date_added'] = pd.to_datetime(df['date_added'], errors='coerce')

df['added_year'] = df['date_added'].dt.year

df['added_year'].value_counts().sort_index().plot()

plt.title('Content Added to Netflix by Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles')

plt.show()
```

### Interpretation
Flat Early Baseline: For the first several years shown on the graph, the volume of newly added titles remained completely flat and close to zero, representing a period when content acquisition on the platform was minimal.

Initial Growth Spark: A slight upward trend begins to take shape in the mid-2010s, marking the initial phase of the platform's expansion into building a larger digital library.

Rapid Acceleration: Following this initial spark, the line transitions into a remarkably steep, rapid climb. This indicates an aggressive, multi-year content acquisition and production boom where massive volumes of content were uploaded annually.

Peak and Recent Decline: After reaching its highest point, the graph takes a distinct downward turn in the final years. This recent drop-off could point to a strategic shift toward quality over quantity, global content production slowdowns, or an incomplete data tracking window for the final calendar year.

# 6. Key Findings

Content Type Imbalance: The platform’s library is heavily dominated by movies, with the number of films more than double the number of TV shows. This indicates a strong preference for standalone content over episodic series.

Geographic Centralization: Content production is largely concentrated in the United States, with India as the second major contributor. However, a significant portion of the dataset is labeled as “Unknown,” indicating incomplete geographic metadata.

Target Audience Focus: The platform primarily targets mature and teenage audiences, with TV-MA and TV-14 being the most frequent ratings. Content designed for young children represents only a small portion of the overall catalog.

Modern Content Dominance: The dataset is largely composed of modern titles, with a sharp increase in content released during the 2010s. Older 20th-century content remains limited, showing a strong focus on recent productions.

Growth Trend Over Time: Content additions increased rapidly during the 2010s, reaching a peak before declining in recent years. This reflects a shift from aggressive expansion to slower growth.

Strong International Presence: International content forms a key part of the catalog, with categories such as “International Movies” and “Dramas” ranking highly in volume.

# 7. Actionable Insights
Improve Content Balance: Increase investment in TV series to improve long-term user engagement and retention, as episodic content typically drives sustained viewing compared to standalone films.

Diversify Content Sources: Reduce reliance on the United States by expanding content production and licensing in emerging international markets, leveraging the strong performance of international categories.

Strengthen Data Quality Controls: Implement stricter data validation during ingestion to prevent errors such as runtime values (e.g., “74 min”) being incorrectly stored in rating fields.

Enhance Metadata Completeness: Enforce mandatory geographic tagging for all content entries to reduce the number of “Unknown” classifications and improve analytical accuracy.

Optimize Content Strategy: Shift focus on high-quality original productions, particularly in mature and teen genres, in response to the recent slowdown in overall content additions.

# 8. Conclusion
The results revealed that Netflix's catalog is dominated by movies, heavily influenced by content from the United States, and primarily targeted toward mature and teenage audiences. Additionally, the platform experienced substantial content growth throughout the 2010s before showing signs of slower expansion in recent years.

Based on the findings, several recommendations were proposed, including increasing investment in TV series, expanding international content production, improving metadata quality, and strengthening data governance practices.
